##Упражнение 7.2
Реализация быстрого ДФП

```python
#ДПФ из книги
def dft(ys):
    N = len(ys)
    ts = np.arange(N) / N
    freqs = np.arange(N)
    args = np.outer(ts, freqs)
    M = np.exp(1j * np.pi * 2 * args)
    amps = M.conj().transpose().dot(ys)
    return amps
```

Быстрое преобразование Фурье (БПФ) предполагает алгоритм с разделением на четные и нечетные элементы и рекурсивным вычислением

```python
#БПФ
def fft(ys):
    N = len(ys)
    if N == 1:
        return ys

    He = fft(ys[::2])
    Ho = fft(ys[1::2])

    ns = np.arange(N)
    W = np.exp(-1j * np.pi * 2 * ns / N)

    return np.tile(He, 2) + W * np.tile(Ho, 2)
```
